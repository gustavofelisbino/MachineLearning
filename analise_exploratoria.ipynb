```python
# ==============================================================================
# 1. IMPORTAÇÃO DAS BIBLIOTECAS
# ==============================================================================
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Configuração visual
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (12, 6)

# ==============================================================================
# 2. CARREGAMENTO DOS DADOS (Certifique-se que os CSVs estão na pasta)
# ==============================================================================
# Carregando apenas as tabelas essenciais para recomendação de produtos
try:
    df_items = pd.read_csv('olist_order_items_dataset.csv')
    df_products = pd.read_csv('olist_products_dataset.csv')
    df_orders = pd.read_csv('olist_orders_dataset.csv')
    df_reviews = pd.read_csv('olist_order_reviews_dataset.csv')
    print("Dados carregados com sucesso!")
except FileNotFoundError:
    print("ERRO: Arquivos CSV não encontrados. Baixe o dataset da Olist no Kaggle.")

# ==============================================================================
# 3. TRATAMENTO E MERGE (UNIFICAÇÃO) DAS TABELAS
# ==============================================================================
# Unindo Itens com Produtos (para ter a categoria)
df_merged = df_items.merge(df_products, on='product_id', how='left')

# Unindo com Reviews (para saber se o cliente gostou)
df_merged = df_merged.merge(df_reviews, on='order_id', how='left')

print(f"Dimensões do Dataset Unificado: {df_merged.shape}")
display(df_merged.head())

# ==============================================================================
# 4. ANÁLISE EXPLORATÓRIA (EDA)
# ==============================================================================

# 4.1. Curva ABC: Top 10 Categorias mais vendidas
plt.figure(figsize=(12, 6))
top_categories = df_merged['product_category_name'].value_counts().head(10)
sns.barplot(x=top_categories.values, y=top_categories.index, palette='viridis')
plt.title('Top 10 Categorias de Produtos Mais Vendidos')
plt.xlabel('Quantidade de Itens')
plt.show()

# 4.2. Distribuição de Preços (Histograma)
# Vamos filtrar preços menores que 500 para visualizar melhor a distribuição
plt.figure(figsize=(10, 5))
sns.histplot(df_merged[df_merged['price'] < 500]['price'], bins=50, kde=True, color='blue')
plt.title('Distribuição de Preços dos Produtos (até R$ 500)')
plt.xlabel('Preço (R$)')
plt.show()

# 4.3. Análise de Avaliações (Review Score)
plt.figure(figsize=(8, 5))
sns.countplot(x='review_score', data=df_merged, palette='magma')
plt.title('Distribuição das Notas de Avaliação (1 a 5)')
plt.xlabel('Nota')
plt.ylabel('Contagem')
plt.show()

# 4.4. Diagnóstico de Valores Nulos
print("\n--- Contagem de Valores Nulos ---")
print(df_merged.isnull().sum())
```
